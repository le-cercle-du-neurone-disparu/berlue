```python
import pandas as pd

# Charger le fichier QA
url = "https://raw.githubusercontent.com/RUCAIBox/HaluEval/main/data/qa_data.json"
df = pd.read_json(url, lines=True)
print(df.head())
```

```python
df_melted = df.melt(
    id_vars=["knowledge", "question"],
    value_vars=["right_answer", "hallucinated_answer"],
    var_name="answer_type",
    value_name="answer",
    ignore_index=False
)


df_melted = df_melted.sort_index()


df_melted["hallucinated"] = df_melted["answer_type"] == "hallucinated_answer"
df_melted = df_melted.drop(columns=["answer_type"]).reset_index(drop=True)
```
